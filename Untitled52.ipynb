```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load Fashion MNIST dataset

(x_train, y_train), (x_test, y_test) = keras.datasets.fashion_mnist.load_data()

print("Training images shape:", x_train.shape)
print("Testing images shape:", x_test.shape)
```

```python
plt.figure(figsize=(10, 4))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.title(str(y_train[i]))
    plt.axis("off")

plt.show()
```

```python
# Normalize pixel values

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

print("Training data shape:", x_train.shape)
print("Minimum value:", x_train.min())
print("Maximum value:", x_train.max())
```

```python
# Encoder settings
latent_dim = 2

# Input layer
encoder_inputs = keras.Input(shape=(28, 28))

# Flatten 28x28 image into a vector
x = layers.Flatten()(encoder_inputs)

# Hidden layer
x = layers.Dense(256, activation="relu")(x)

# Mean and variance of latent space
z_mean = layers.Dense(latent_dim)(x)
z_log_var = layers.Dense(latent_dim)(x)

print("Encoder created successfully")
```

```python
class Sampling(layers.Layer):
    def call(self, inputs):
        z_mean, z_log_var = inputs

        batch = tf.shape(z_mean)[0]
        dim = tf.shape(z_mean)[1]

        epsilon = tf.random.normal(shape=(batch, dim))

        return z_mean + tf.exp(0.5 * z_log_var) * epsilon
```

```python
# Decoder input
latent_inputs = keras.Input(shape=(latent_dim,))

# Hidden layer
x = layers.Dense(256, activation="relu")(latent_inputs)

# Output layer: 28 × 28 = 784 pixels
x = layers.Dense(28 * 28, activation="sigmoid")(x)

# Reshape back to image
decoder_outputs = layers.Reshape((28, 28))(x)

# Create decoder model
decoder = keras.Model(latent_inputs, decoder_outputs, name="decoder")

print("Decoder created successfully")
```

```python
z = Sampling()([z_mean, z_log_var])

encoder = keras.Model(
    encoder_inputs,
    [z_mean, z_log_var, z],
    name="encoder"
)

encoder.summary()
```

```python
class VAE(keras.Model):
    def __init__(self, encoder, decoder, **kwargs):
        super().__init__(**kwargs)
        self.encoder = encoder
        self.decoder = decoder

    def call(self, inputs):
        z_mean, z_log_var, z = self.encoder(inputs)
        reconstructed = self.decoder(z)
        return reconstructed

vae = VAE(encoder, decoder)

print("VAE created successfully")
```

```python
class VAE(keras.Model):
    def __init__(self, encoder, decoder, **kwargs):
        super().__init__(**kwargs)
        self.encoder = encoder
        self.decoder = decoder

    def train_step(self, data):
        if isinstance(data, tuple):
            data = data[0]

        with tf.GradientTape() as tape:
            z_mean, z_log_var, z = self.encoder(data)
            reconstruction = self.decoder(z)

            # Reconstruction loss
            reconstruction_loss = tf.reduce_mean(
                tf.reduce_sum(
                    keras.losses.binary_crossentropy(data, reconstruction),
                    axis=(1)
                )
            )

            # KL divergence loss
            kl_loss = -0.5 * tf.reduce_mean(
                tf.reduce_sum(
                    1 + z_log_var - tf.square(z_mean) - tf.exp(z_log_var),
                    axis=1
                )
            )

            total_loss = reconstruction_loss + kl_loss

        grads = tape.gradient(
            total_loss,
            self.trainable_weights
        )

        self.optimizer.apply_gradients(
            zip(grads, self.trainable_weights)
        )

        return {
            "loss": total_loss,
            "reconstruction_loss": reconstruction_loss,
            "kl_loss": kl_loss,
        }
```

```python
vae = VAE(encoder, decoder)

vae.compile(
    optimizer=keras.optimizers.Adam()
)

print("VAE compiled successfully")
```

```python
vae.fit(
    x_train,
    epochs=10,
    batch_size=128
)
```

```python
# Generate 10 random latent vectors
z_samples = np.random.normal(size=(10, latent_dim))

# Generate images using the decoder
generated_images = decoder.predict(z_samples)

# Display generated images
plt.figure(figsize=(15, 3))

for i in range(10):
    plt.subplot(1, 10, i + 1)
    plt.imshow(generated_images[i], cmap="gray")
    plt.axis("off")

plt.show()
```
