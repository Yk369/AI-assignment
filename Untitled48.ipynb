```python
n = int(input("Enter a number: "))

fact = 1
for i in range(1, n + 1):
    fact *= i

print("Factorial =", fact)
```

```python
n = int(input("Enter n: "))

sum = 0
for i in range(1, n + 1):
    sum += i

print("Sum =", sum)
```

```python
A = [[1, 2, 3],
    [4, 5, 6]]
B = [[7, 8, 9],
    [10, 11, 12]]
C = []

for i in range(len(A)):
    row = []
    for j in range(len(A[0])):
        row.append(A[i][j] + B[i][j])
    C.append(row)

print("Sum of matrices:")
for row in C:
    print(row)
```
