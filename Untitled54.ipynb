```python
import cv2
import numpy as np
from skimage import data
from google.colab.patches import cv2_imshow

img_rgb = data.coffee()
img = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR)

image = img.astype(np.float32)

B, G, R = cv2.split(image)

mean_B = np.mean(B)
mean_G = np.mean(G)
mean_R = np.mean(R)


gray_mean = (mean_B + mean_G + mean_R) / 3

gain_B = gray_mean / mean_B
gain_G = gray_mean / mean_G
gain_R = gray_mean / mean_R

B_corrected = B * gain_B
G_corrected = G * gain_G
R_corrected = R * gain_R

B_corrected = np.clip(B_corrected, 0, 255)
G_corrected = np.clip(G_corrected, 0, 255)
R_corrected = np.clip(R_corrected, 0, 255)

corrected = cv2.merge([B_corrected, G_corrected, R_corrected]).astype(np.uint8)

print("Original Image")
cv2_imshow(img)
print("Color-Corrected Image")
cv2_imshow(corrected)
print("Original means:")
print(mean_B, mean_G, mean_R)
print("Channel gains:")
print(gain_B, gain_G, gain_R)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
image = cv2.imread("/content/ChatGPT Image Sep 2, 2026, 09_46_44 AM.png")

# Split into B, G, R channels
B, G, R = cv2.split(image)

# Find highest and lowest pixel values
print("RED CHANNEL")
print("Highest value:", np.max(R))
print("Lowest value:", np.min(R))

print("\nGREEN CHANNEL")
print("Highest value:", np.max(G))
print("Lowest value:", np.min(G))

print("\nBLUE CHANNEL")
print("Highest value:", np.max(B))
print("Lowest value:", np.min(B))

# Find which channel has the highest maximum value
max_values = {
    "Red": np.max(R),
    "Green": np.max(G),
    "Blue": np.max(B)
}

# Find which channel has the lowest minimum value
min_values = {
    "Red": np.min(R),
    "Green": np.min(G),
    "Blue": np.min(B)
}

highest_channel = max(max_values, key=max_values.get)
lowest_channel = min(min_values, key=min_values.get)

print("\nOverall highest channel:", highest_channel)
print("Value:", max_values[highest_channel])

print("Overall lowest channel:", lowest_channel)
print("Value:", min_values[lowest_channel])

# Display channels
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(R, cmap="gray")
plt.title("Red Channel")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(G, cmap="gray")
plt.title("Green Channel")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(B, cmap="gray")
plt.title("Blue Channel")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read image
image = cv2.imread("/content/ChatGPT Image Sep 2, 2026, 09_46_44 AM.png")

# Convert image from BGR to float
image = image.astype(np.float32)

# Calculate average of each color channel
B_avg = np.mean(image[:, :, 0])
G_avg = np.mean(image[:, :, 1])
R_avg = np.mean(image[:, :, 2])

# Calculate overall gray average
gray_avg = (B_avg + G_avg + R_avg) / 3

# Calculate scaling factors
B_scale = gray_avg / B_avg
G_scale = gray_avg / G_avg
R_scale = gray_avg / R_avg

# Apply scaling
image[:, :, 0] *= B_scale
image[:, :, 1] *= G_scale
image[:, :, 2] *= R_scale

# Keep pixel values between 0 and 255
image = np.clip(image, 0, 255)

# Convert back to uint8
image = image.astype(np.uint8)

# Save corrected image
cv2.imwrite("gray_world_output.jpg", image)

# Display result
cv2_imshow(image)
```
